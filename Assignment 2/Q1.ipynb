```python
def first_fit(blocks, processes):
    return allocate(blocks, processes, "First Fit")


def best_fit(blocks, processes):
    return allocate(blocks, processes, "Best Fit")


def worst_fit(blocks, processes):
    return allocate(blocks, processes, "Worst Fit")


def allocate(blocks, processes, strategy):
    memory = blocks.copy()
    allocation = [-1] * len(processes)

    for i, process in enumerate(processes):
        suitable = []

        for j in range(len(memory)):
            if memory[j] >= process:
                suitable.append(j)

        if suitable:
            if strategy == "First Fit":
                index = suitable[0]
            elif strategy == "Best Fit":
                index = min(suitable, key=lambda j: memory[j])
            else:
                index = max(suitable, key=lambda j: memory[j])

            allocation[i] = index
            memory[index] -= process

    print("\n", strategy)
    print("-" * 48)
    print("Process\tSize\tBlock\tStatus")

    for i, process in enumerate(processes):
        if allocation[i] != -1:
            block = allocation[i] + 1
            status = "Allocated"
        else:
            block = "-"
            status = "Not Allocated"

        print(f"P{i+1}\t{process}\t{block}\t{status}")

    allocated = allocation.count_not_allocated() if False else sum(
        1 for x in allocation if x != -1
    )
    free_memory = sum(memory)
    used_memory = sum(blocks) - free_memory
    utilization = (used_memory / sum(blocks)) * 100

    print("Successfully allocated:", allocated)
    print("Remaining free memory:", free_memory)
    print("Memory utilization: {:.2f}%".format(utilization))

    return allocated, free_memory, utilization


blocks = [100, 500, 200, 300, 600]
processes = [212, 417, 112, 426]

print("Memory Blocks:", blocks)
print("Processes:", processes)

results = {}

for strategy in [first_fit, best_fit, worst_fit]:
    name = strategy.__name__.replace("_", " ").title()
    results[name] = strategy(blocks, processes)

print("\nComparison of Strategies")
print("-" * 65)
print("Strategy\tAllocated\tFree Memory\tUtilization")

for name, result in results.items():
    print("{}\t{}\t\t{}\t\t{:.2f}%".format(
        name, result[0], result[1], result[2]
    ))
```
