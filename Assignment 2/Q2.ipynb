```python

def fifo(pages, capacity):
    frames = []
    faults = hits = 0
    pointer = 0

    print("\nFIFO Page Replacement")
    print("Page\tFrames\t\tStatus")

    for page in pages:
        if page in frames:
            hits += 1
            status = "Hit"
        else:
            faults += 1
            status = "Fault"

            if len(frames) < capacity:
                frames.append(page)
            else:
                frames[pointer] = page
                pointer = (pointer + 1) % capacity

        print(f"{page}\t{frames}\t{status}")

    return faults, hits


def lru(pages, capacity):
    frames = []
    recent = []
    faults = hits = 0

    print("\nLRU Page Replacement")
    print("Page\tFrames\t\tStatus")

    for page in pages:
        if page in frames:
            hits += 1
            status = "Hit"
            recent.remove(page)
            recent.append(page)
        else:
            faults += 1
            status = "Fault"

            if len(frames) < capacity:
                frames.append(page)
            else:
                old_page = recent.pop(0)
                frames[frames.index(old_page)] = page

            recent.append(page)

        print(f"{page}\t{frames}\t{status}")

    return faults, hits


def optimal(pages, capacity):
    frames = []
    faults = hits = 0

    print("\nOptimal Page Replacement")
    print("Page\tFrames\t\tStatus")

    for i, page in enumerate(pages):
        if page in frames:
            hits += 1
            status = "Hit"
        else:
            faults += 1
            status = "Fault"

            if len(frames) < capacity:
                frames.append(page)
            else:
                future = pages[i + 1:]
                distances = []

                for frame in frames:
                    if frame in future:
                        distances.append(future.index(frame))
                    else:
                        distances.append(float("inf"))

                replace_index = distances.index(max(distances))
                frames[replace_index] = page

        print(f"{page}\t{frames}\t{status}")

    return faults, hits


pages = [7, 0, 1, 2, 0, 3, 0, 4, 2, 3, 0, 3, 2]
capacity = 3

print("Page Reference String:", pages)
print("Number of Frames:", capacity)

results = {
    "FIFO": fifo(pages, capacity),
    "LRU": lru(pages, capacity),
    "Optimal": optimal(pages, capacity)
}

print("\nComparison of Algorithms")
print("Algorithm\tPage Faults\tPage Hits")

for name, result in results.items():
    print(f"{name}\t\t{result[0]}\t\t{result[1]}")
```
