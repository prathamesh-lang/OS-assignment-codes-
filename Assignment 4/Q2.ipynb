```python

def fcfs(req, head):
    order = req[:]
    return order

def sstf(req, head):
    req = req[:]
    order = []
    while req:
        nearest = min(req, key=lambda x: abs(x - head))
        order.append(nearest)
        req.remove(nearest)
        head = nearest
    return order

def scan(req, head, size):
    left = sorted([x for x in req if x < head], reverse=True)
    right = sorted([x for x in req if x >= head])
    return right + ([size - 1] if right else []) + left

def cscan(req, head, size):
    left = sorted([x for x in req if x < head])
    right = sorted([x for x in req if x >= head])
    return right + ([size - 1, 0] if left else []) + left

def movement(order, head):
    total = 0
    for track in order:
        total += abs(track - head)
        head = track
    return total

requests = [82, 170, 43, 140, 24, 16, 190]
head = 50
disk_size = 200

algorithms = {
    "FCFS": fcfs(requests, head),
    "SSTF": sstf(requests, head),
    "SCAN": scan(requests, head, disk_size),
    "C-SCAN": cscan(requests, head, disk_size)
}

for name, order in algorithms.items():
    print(name, "Order:", order)
    print("Total head movement:", movement(order, head))
    print()
```
