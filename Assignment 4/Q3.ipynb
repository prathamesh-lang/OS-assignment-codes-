```python

import ipaddress

def validate_ip(ip):
    try:
        return ipaddress.IPv4Address(ip)
    except ipaddress.AddressValueError:
        return None

ip1 = input("Enter first host IP address: ")
ip2 = input("Enter second host IP address: ")
mask = input("Enter subnet mask (e.g. 255.255.255.0): ")

try:
    host1 = validate_ip(ip1)
    host2 = validate_ip(ip2)

    if host1 is None or host2 is None:
        print("Invalid IP address")
    else:
        network1 = ipaddress.IPv4Network(
            f"{ip1}/{mask}", strict=False)
        network2 = ipaddress.IPv4Network(
            f"{ip2}/{mask}", strict=False)

        print("First IP:", host1)
        print("Second IP:", host2)
        print("Subnet:", network1.netmask)
        print("Network address 1:", network1.network_address)
        print("Network address 2:", network2.network_address)

        if network1.network_address == network2.network_address:
            print("Both hosts belong to the same subnet.")
        else:
            print("Hosts belong to different subnets.")

except ValueError:
    print("Invalid subnet mask.")
```
